```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
#graph threshold were based on IH and IL which are Input High and Input Low within the csv files
#in a nor gate we have: 
# 0 1 = 0
# 0 0 = 1
# 1 1 = 0 
# 1 0 = 0
# this means if x < IL output = 1 and if x > IH output = 0
# so when x < IL output = high = green 
# and when x > IH output = low = blue
# and when IH < x < IL output = transition = yellow
file = pd.read_csv("parsed_Eco1C1G1T1.csv")
fig, axes = plt.subplots(5, 4, figsize=(20, 20))
x = np.logspace(-3, 2, 400)

for index, row in file.iterrows():
    axis = axes.flat[index]
    #NOR gate equation: y = ymin + (ymax - ymin) / (1 + (x/K)^n)
    y = row["ymin"] + (row["ymax"] - row["ymin"]) / (1.0 + (x / row["K"]) ** row["n"])
    #highlighting the different regions based on IL and IH 
    axis.axvspan(x.min(), row["IL"], color= "green", alpha=0.25, label="Low input / ON" if index == 0 else None)
    axis.axvspan(row["IL"], row["IH"], color= "yellow", alpha=0.25,label="Transition" if index == 0 else None)
    axis.axvspan(row["IH"], x.max(), color= "blue", alpha=0.25,label="High input / OFF" if index == 0 else None)
    axis.plot(x, y, color="black", linewidth=1.5, zorder=2)
    axis.set_xscale("log")
    axis.set_box_aspect(1)
    axis.set_title(f"{row['name']}\nK={row['K']}, n={row['n']}")
    axis.set_xlabel(f"Input")
    axis.set_ylabel("Protein Expression")

fig.legend(*axes.flat[0].get_legend_handles_labels(), loc="upper center", ncol=3, frameon=False)
plt.tight_layout(rect=(0, 0, 1, 0.96))
plt.show()
```
